##Insight questão 10

Ao separar corretamente os países coprodutores, foi observado uma distribuição mais fiel da origem dos títulos.

Os Estados Unidos permanecem como o principal produtor de conteúdo disponível na plataforma, seguidos por Índia e Reino Unido.

A análise também evidencia que muitos títulos são resultado de coproduções internacionais, o que reforça a importância de tratar corretamente colunas com múltiplos valores em projetos de análise de dados.

```python
#Questão 10 - Código
#Copiar df para manipulação
countries_df = df.copy()

#Remover valores ausentes
countries_df = countries_df.dropna(subset=["country"])
#Checar
countries_df["country"].isna().sum()

#Separar valores
countries_df["country"] = (
    countries_df["country"]
    .str.split(", ")
)
#Conferir
countries_df["country"].head()

#Transformar itens em linhas
countries_df = countries_df.explode("country")
#Conferir
countries_df.head(10)

#Contagem
country_count = (
    countries_df["country"]
    .value_counts()
)

#Top 15
top_countries = country_count.head(15)

top_countries

#Percentual
country_percentage = (
    countries_df["country"]
    .value_counts(normalize=True)
    .mul(100)
    .round(2)
)

country_percentage.head(15)

#Gráfico
plt.figure(figsize=(12, 8))

ax = sns.barplot(
    x=top_countries.values,
    y=top_countries.index,
    hue=top_countries.index,
    legend=False
)

plt.title("Top 15 Producing Countries")

plt.xlabel("Number of Productions")

plt.ylabel("Country")

for container in ax.containers:
    ax.bar_label(
        container,
        padding=3
    )

plt.tight_layout()

plt.savefig(
    "../images/top_producing_countries_corrected.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
#Questão 6 - Código
#Preparar dados
months = [
    "January",
    "February",
    "March",
    "April",
    "May",
    "June",
    "July",
    "August",
    "September",
    "October",
    "November",
    "December"
]

df["month_added"] = pd.Categorical(
    df["month_added"],
    categories=months,
    ordered=True
)

#Contagem
titles_month = (
    df["month_added"]
    .value_counts()
    .sort_index()
)

titles_month

#Gráfico
plt.figure(figsize=(12,6))

ax = sns.barplot(
    x=titles_month.index,
    y=titles_month.values
)

plt.xticks(rotation=45)

plt.title("Titles Added by Month")

plt.xlabel("Month")

plt.ylabel("Number of Titles")

for container in ax.containers:
    ax.bar_label(container)

plt.tight_layout()

plt.savefig(
    "../images/titles_added_by_month.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
